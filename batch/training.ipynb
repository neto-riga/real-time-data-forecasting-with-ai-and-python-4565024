```python
Y_train = feature_store.fetch_data_from_store(X_train.index.min(), yaml_file_path, targets = True)[:X_train.index.max()]
Y_train
```

```python
# turn into a list
y_train_ls= []
for target in Y_train.columns:
  y_train_ls.append(Y_train[target])

y_train_ls
```

```python
# Loop through targets - train one model for each target variable
cv_results_ls = []
for y_train in y_train_ls:

  # Prepare the DMatrix which is required by XGBoost
  dtrain = xgb.DMatrix(data=X_train, label=y_train)

  # Define XGBoost parameters
  params = {
    #'max_depth': 6,
    #'min_child_weight': 1,
    #'eta': 0.5,
    #'subsample': 1,
    #'colsample_bytree': 1,
    'objective': 'reg:squarederror',
    'eval_metric': 'rmse'
  }

  # Perform cross-validation
  cv_results = xgb.cv(
      params=params,
      dtrain=dtrain,
      num_boost_round=100,
      nfold = 4,
      early_stopping_rounds=10,
      metrics='rmse',
      as_pandas=True,
      seed=123
  )
  cv_results_ls.append(cv_results)
  # Show the last mean RMSE as a measure of final performance
  print(f"Last mean RMSE: {cv_results['test-rmse-mean'][-1:]}")
  
```

```python
from sklearn.model_selection import TimeSeriesSplit
import xgboost as xgb
import numpy as np

params = {
    'eta': 0.05,
    'max_depth': 3,
    'min_child_weight': 5,
    'subsample': 0.8,
    'colsample_bytree': 0.8,
    'tree_method': 'hist',
    'verbosity': 2
}

# Build time-aware folds once, aligned to X_train row order
tscv = TimeSeriesSplit(n_splits=2)  # optionally: TimeSeriesSplit(n_splits=4, gap=2)
folds = list(tscv.split(X_train))   # returns (train_idx, test_idx) by POSITION

cv_results_ls = []
for h, y_train in enumerate(y_train_ls, start=1):
    dtrain = xgb.DMatrix(X_train.copy(), label=y_train)

    # Use the time-ordered folds; this overrides random nfold behavior
    cv_results = xgb.cv(
        params=params,
        dtrain=dtrain,
        folds=folds,                  # <-- key line
        num_boost_round=500,
        # early_stopping_rounds=30,
        metrics='rmse',
        as_pandas=True,
        seed=123
    )
    cv_results_ls.append(cv_results)
    print(f"H+{h} last mean RMSE: {cv_results['test-rmse-mean'].iloc[-1]}")
    print(len(cv_results))
```

```python
import matplotlib.pyplot as plt

# Plot RMSE through iterations with different colors for each line

plt.figure(figsize=(10, 6))

for idx, cv_results in enumerate(cv_results_ls):
    plt.plot(cv_results['test-rmse-mean'], label=f'CV Run {idx+1}')

plt.title('Test RMSE Through Iterations')
plt.xlabel('Number of Boosting Rounds')
plt.ylabel('Mean RMSE')
plt.legend()
plt.show()
```
